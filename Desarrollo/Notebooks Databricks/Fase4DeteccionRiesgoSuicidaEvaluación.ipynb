###**Matriz Confusión Naive Bayes (Texto)**

En este apartado se medirá el desempeño del modelo Naive Bayes basado en texto, utilizando la matriz de confusión para comparar las clases reales y predichas. A partir de esta matriz se analizarán los errores de clasificación y se calcularán métricas por clase como accuracy, precision, recall y F1-score, con el fin de evaluar la capacidad del modelo para identificar correctamente casos de riesgo.

```python
# Extraer labels reales y predicciones
cm_df_nb = nb_text_predictions.select(
    col("class").cast("int").alias("y_true_nb"),
    col("prediction").cast("int").alias("y_pred_nb")
)

# Convertir a Pandas
pdf_cm_nb = cm_df_nb.toPandas()

# Etiquetas únicas
labels_nb = sorted(pdf_cm_nb["y_true_nb"].unique())

# Matriz de confusión
cm_nb = confusion_matrix(
    pdf_cm_nb["y_true_nb"],
    pdf_cm_nb["y_pred_nb"],
    labels=labels_nb
)

# Normalización por filas
cm_norm_nb = cm_nb.astype("float") / cm_nb.sum(axis=1)[:, np.newaxis]

# DataFrame para graficar
cm_plot_df_nb = pd.DataFrame(
    cm_nb,
    index=[f"Real {l}" for l in labels_nb],
    columns=[f"Pred {l}" for l in labels_nb]
)

# Texto (conteo + porcentaje)
text_cm_nb = [
    [f"{cm_nb[i, j]}<br>({cm_norm_nb[i, j]:.1%})" for j in range(cm_nb.shape[1])]
    for i in range(cm_nb.shape[0])
]

# Paleta morada – Salud Mental / Riesgo
purple_risk_scale = [
    "#f4f1fa",
    "#e0d6f5",
    "#c1a7e2",
    "#9b6bd6",
    "#6a3fa0",
    "#432874"
]

# Gráfico Plotly
fig_cm_nb = px.imshow(
    cm_plot_df_nb,
    text_auto=False,
    color_continuous_scale=purple_risk_scale,
    aspect="auto",
    title="Matriz de Confusión – Naive Bayes (Texto)"
)

fig_cm_nb.update_traces(
    text=text_cm_nb,
    texttemplate="%{text}",
    textfont_size=20,
    textfont_color="white"
)

fig_cm_nb.update_layout(
    title_x=0.5,
    xaxis_title="Clase Predicha",
    yaxis_title="Clase Real",
    font=dict(
        family="Algerian, DejaVu Sans, sans-serif",
        size=18
    ),
    height=600,
    width=720,
    margin=dict(l=90, r=90, t=120, b=70),
    coloraxis_colorbar=dict(title="Frecuencia")
)

fig_cm_nb.show()

# Métricas por clase
rows_nb = []
total_samples_nb = cm_nb.sum()

for i, label in enumerate(labels_nb):
    TP_nb = cm_nb[i, i]
    FN_nb = cm_nb[i, :].sum() - TP_nb
    FP_nb = cm_nb[:, i].sum() - TP_nb
    TN_nb = total_samples_nb - (TP_nb + FP_nb + FN_nb)

    precision_nb = TP_nb / (TP_nb + FP_nb) if (TP_nb + FP_nb) > 0 else 0.0
    recall_nb = TP_nb / (TP_nb + FN_nb) if (TP_nb + FN_nb) > 0 else 0.0
    f1_nb = (
        2 * precision_nb * recall_nb / (precision_nb + recall_nb)
        if (precision_nb + recall_nb) > 0
        else 0.0
    )
    accuracy_nb = (TP_nb + TN_nb) / total_samples_nb
    registros_nb = int(cm_nb[i, :].sum())

    rows_nb.append({
        "Class": int(label),
        "Accuracy": builtins.round(float(accuracy_nb), 4),
        "Precision": builtins.round(float(precision_nb), 4),
        "Recall": builtins.round(float(recall_nb), 4),
        "F1-score": builtins.round(float(f1_nb), 4),
        "Rows": registros_nb
    })

# Mostrar tabla de métricas
metrics_by_class_nb = pd.DataFrame(rows_nb)
display(metrics_by_class_nb)
```

###**Sensibilidad Modelo Según Umbral - Naive Bayes (Texto)**

En este apartado se medirá la sensibilidad (recall) del modelo Naive Bayes basado en texto en función del umbral de decisión, analizando cómo varía la detección de casos de riesgo al modificar dicho umbral. Esta evaluación permitirá identificar configuraciones que prioricen la detección de riesgo, aun cuando esto implique un incremento en los falsos positivos.

```python
# Configuración de fuentes
rcParams["font.family"] = "DejaVu Sans"
rcParams["axes.titlesize"] = 18
rcParams["axes.labelsize"] = 14

# Extraer y_true y Probabilidad (Clase Positiva = 1)
recall_df_nb = nb_text_predictions.select(
    col("class").cast("int").alias("y_true_nb"),
    col("probability").alias("prob_nb")
)

# Convertir a Pandas
pdf_recall_nb = recall_df_nb.toPandas()

# Extraer etiquetas reales y probabilidades predichas de la clase positiva
y_true_nb = pdf_recall_nb["y_true_nb"].values
y_score_nb = pdf_recall_nb["prob_nb"].apply(lambda x: float(x[1])).values

# Calcular recall para distintos threshold
thresholds_nb = np.linspace(0.0, 1.0, 101)
recall_values_nb = []

for t in thresholds_nb:
    y_pred_t_nb = (y_score_nb >= t).astype(int)
    recall_t_nb = recall_score(y_true_nb, y_pred_t_nb, zero_division=0)
    recall_values_nb.append(recall_t_nb)

# Threshold aproximado donde recall >= 90%
threshold_obj_nb = thresholds_nb[
    np.where(np.array(recall_values_nb) >= 0.90)[0][-1]
]

# Gráficar recall vs threshold
fig, ax = plt.subplots(figsize=(10.5, 7.5))

# Curva principal
ax.plot(
    thresholds_nb,
    recall_values_nb,
    linewidth=3.5,
    color="#6a3fa0",
    solid_capstyle="round",
    label="Recall (Clase Riesgo)"
)

# Sombreado suave
ax.fill_between(
    thresholds_nb,
    recall_values_nb,
    color="#6a3fa0",
    alpha=0.12
)

# Línea objetivo
ax.axhline(
    y=0.90,
    linestyle="--",
    linewidth=2,
    color="#9b6bd6",
    label="Recall Objetivo (90%)"
)

# Punto óptimo
ax.scatter(
    threshold_obj_nb,
    0.90,
    s=90,
    color="#432874",
    zorder=5
)

# Títulos y etiquetas
ax.set_title(
    "Modelo Naive Bayes (Texto)\n"
    "\nSensibilidad Modelo Según Umbral",
    fontsize=20,
    pad=20
)

ax.set_xlabel("\nUmbral Decisión", fontsize=14)
ax.set_ylabel("Recall (Sensibilidad)", fontsize=14)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)

# Grid y leyenda
ax.grid(
    linestyle=":",
    linewidth=0.8,
    alpha=0.4
)

ax.legend(
    fontsize=12,
    frameon=False
)

# Anotación explicativa sobre efecto del umbral bajo
ax.annotate(
    "Umbral Bajo → Mayor Detección de Riesgo\n"
    "(A Costa de Más Falsos Positivos)",
    xy=(0.22, 0.90),
    xytext=(0.48, 0.55),
    arrowprops=dict(
        arrowstyle="->",
        color="#6a3fa0",
        lw=2
    ),
    fontsize=12,
    bbox=dict(
        boxstyle="round,pad=0.4",
        fc="white",
        ec="#6a3fa0",
        alpha=0.85
    )
)

# Mostrar gráfico
plt.show()
```

```python
# Extraer labels reales y predicciones
cm_df_lr_text_ner = lr_predictions_text_ner.select(
    col("class").cast("int").alias("y_true_lr_text_ner"),
    col("prediction").cast("int").alias("y_pred_lr_text_ner")
)

# Convertir a Pandas
pdf_cm_lr_text_ner = cm_df_lr_text_ner.toPandas()

# Etiquetas únicas
labels_lr_text_ner = sorted(pdf_cm_lr_text_ner["y_true_lr_text_ner"].unique())

# Matriz de confusión
cm_lr_text_ner = confusion_matrix(
    pdf_cm_lr_text_ner["y_true_lr_text_ner"],
    pdf_cm_lr_text_ner["y_pred_lr_text_ner"],
    labels=labels_lr_text_ner
)

# Normalización por filas (Recall implícito)
cm_norm_lr_text_ner = (
    cm_lr_text_ner.astype("float") /
    cm_lr_text_ner.sum(axis=1)[:, np.newaxis]
)

# DataFrame para graficar
cm_plot_df_lr_text_ner = pd.DataFrame(
    cm_lr_text_ner,
    index=[f"Real {l}" for l in labels_lr_text_ner],
    columns=[f"Pred {l}" for l in labels_lr_text_ner]
)

# Texto (conteo + porcentaje)
text_cm_lr_text_ner = [
    [
        f"{cm_lr_text_ner[i, j]}<br>({cm_norm_lr_text_ner[i, j]:.1%})"
        for j in range(cm_lr_text_ner.shape[1])
    ]
    for i in range(cm_lr_text_ner.shape[0])
]

# Paleta morada – Salud Mental / Riesgo
purple_risk_scale = [
    "#f4f1fa",
    "#e0d6f5",
    "#c1a7e2",
    "#9b6bd6",
    "#6a3fa0",
    "#432874"
]

# Gráfico Plotly
fig_cm_lr_text_ner = px.imshow(
    cm_plot_df_lr_text_ner,
    text_auto=False,
    color_continuous_scale=purple_risk_scale,
    aspect="auto",
    title="Matriz de Confusión – Logistic Regression (Texto + NER)"
)

fig_cm_lr_text_ner.update_traces(
    text=text_cm_lr_text_ner,
    texttemplate="%{text}",
    textfont_size=20,
    textfont_color="white"
)

fig_cm_lr_text_ner.update_layout(
    title_x=0.5,
    xaxis_title="Clase Predicha",
    yaxis_title="Clase Real",
    font=dict(
        family="Algerian, DejaVu Sans, sans-serif",
        size=18
    ),
    height=600,
    width=720,
    margin=dict(l=90, r=90, t=120, b=70),
    coloraxis_colorbar=dict(title="Frecuencia")
)

fig_cm_lr_text_ner.show()

# Métricas por clase
rows_lr_text_ner = []
total_samples_lr_text_ner = cm_lr_text_ner.sum()

for i, label in enumerate(labels_lr_text_ner):
    TP_lr_text_ner = cm_lr_text_ner[i, i]
    FN_lr_text_ner = cm_lr_text_ner[i, :].sum() - TP_lr_text_ner
    FP_lr_text_ner = cm_lr_text_ner[:, i].sum() - TP_lr_text_ner
    TN_lr_text_ner = (
        total_samples_lr_text_ner -
        (TP_lr_text_ner + FP_lr_text_ner + FN_lr_text_ner)
    )

    precision_lr_text_ner = (
        TP_lr_text_ner / (TP_lr_text_ner + FP_lr_text_ner)
        if (TP_lr_text_ner + FP_lr_text_ner) > 0 else 0.0
    )

    recall_lr_text_ner = (
        TP_lr_text_ner / (TP_lr_text_ner + FN_lr_text_ner)
        if (TP_lr_text_ner + FN_lr_text_ner) > 0 else 0.0
    )

    f1_lr_text_ner = (
        2 * precision_lr_text_ner * recall_lr_text_ner /
        (precision_lr_text_ner + recall_lr_text_ner)
        if (precision_lr_text_ner + recall_lr_text_ner) > 0 else 0.0
    )

    accuracy_lr_text_ner_cls = (
        (TP_lr_text_ner + TN_lr_text_ner) /
        total_samples_lr_text_ner
    )

    registros_lr_text_ner = int(cm_lr_text_ner[i, :].sum())

    rows_lr_text_ner.append({
        "Class": int(label),
        "Accuracy": builtins.round(float(accuracy_lr_text_ner_cls), 4),
        "Precision": builtins.round(float(precision_lr_text_ner), 4),
        "Recall": builtins.round(float(recall_lr_text_ner), 4),
        "F1-score": builtins.round(float(f1_lr_text_ner), 4),
        "Rows": registros_lr_text_ner
    })

# Mostrar tabla de métricas
metrics_by_class_lr_text_ner = pd.DataFrame(rows_lr_text_ner)
display(metrics_by_class_lr_text_ner)
```

```python
# Extraer labels reales y predicciones
cm_df_svm_text_ner = svm_ner_predictions.select(
    col("class").cast("int").alias("y_true_svm_text_ner"),
    col("prediction").cast("int").alias("y_pred_svm_text_ner")
)

# Convertir a Pandas
pdf_cm_svm_text_ner = cm_df_svm_text_ner.toPandas()

# Etiquetas únicas
labels_svm_text_ner = sorted(pdf_cm_svm_text_ner["y_true_svm_text_ner"].unique())

# Matriz de confusión
cm_svm_text_ner = confusion_matrix(
    pdf_cm_svm_text_ner["y_true_svm_text_ner"],
    pdf_cm_svm_text_ner["y_pred_svm_text_ner"],
    labels=labels_svm_text_ner
)

# Normalización por filas (Recall implícito)
cm_norm_svm_text_ner = (
    cm_svm_text_ner.astype("float") /
    cm_svm_text_ner.sum(axis=1)[:, np.newaxis]
)

# DataFrame para graficar
cm_plot_df_svm_text_ner = pd.DataFrame(
    cm_svm_text_ner,
    index=[f"Real {l}" for l in labels_svm_text_ner],
    columns=[f"Pred {l}" for l in labels_svm_text_ner]
)

# Texto (conteo + porcentaje)
text_cm_svm_text_ner = [
    [
        f"{cm_svm_text_ner[i, j]}<br>({cm_norm_svm_text_ner[i, j]:.1%})"
        for j in range(cm_svm_text_ner.shape[1])
    ]
    for i in range(cm_svm_text_ner.shape[0])
]

# Paleta morada – Salud Mental / Riesgo
purple_risk_scale = [
    "#f4f1fa",
    "#e0d6f5",
    "#c1a7e2",
    "#9b6bd6",
    "#6a3fa0",
    "#432874"
]

# Gráfico Plotly
fig_cm_svm_text_ner = px.imshow(
    cm_plot_df_svm_text_ner,
    text_auto=False,
    color_continuous_scale=purple_risk_scale,
    aspect="auto",
    title="Matriz de Confusión – Support Vector Machine (Texto + NER)"
)

fig_cm_svm_text_ner.update_traces(
    text=text_cm_svm_text_ner,
    texttemplate="%{text}",
    textfont_size=20,
    textfont_color="white"
)

fig_cm_svm_text_ner.update_layout(
    title_x=0.5,
    xaxis_title="Clase Predicha",
    yaxis_title="Clase Real",
    font=dict(
        family="Algerian, DejaVu Sans, sans-serif",
        size=18
    ),
    height=600,
    width=720,
    margin=dict(l=90, r=90, t=120, b=70),
    coloraxis_colorbar=dict(title="Frecuencia")
)

fig_cm_svm_text_ner.show()

# Métricas por clase
rows_svm_text_ner = []
total_samples_svm_text_ner = cm_svm_text_ner.sum()

for i, label in enumerate(labels_svm_text_ner):
    TP_svm_text_ner = cm_svm_text_ner[i, i]
    FN_svm_text_ner = cm_svm_text_ner[i, :].sum() - TP_svm_text_ner
    FP_svm_text_ner = cm_svm_text_ner[:, i].sum() - TP_svm_text_ner
    TN_svm_text_ner = (
        total_samples_svm_text_ner -
        (TP_svm_text_ner + FP_svm_text_ner + FN_svm_text_ner)
    )

    precision_svm_text_ner = (
        TP_svm_text_ner / (TP_svm_text_ner + FP_svm_text_ner)
        if (TP_svm_text_ner + FP_svm_text_ner) > 0 else 0.0
    )

    recall_svm_text_ner = (
        TP_svm_text_ner / (TP_svm_text_ner + FN_svm_text_ner)
        if (TP_svm_text_ner + FN_svm_text_ner) > 0 else 0.0
    )

    f1_svm_text_ner = (
        2 * precision_svm_text_ner * recall_svm_text_ner /
        (precision_svm_text_ner + recall_svm_text_ner)
        if (precision_svm_text_ner + recall_svm_text_ner) > 0 else 0.0
    )

    accuracy_svm_text_ner_cls = (
        (TP_svm_text_ner + TN_svm_text_ner) /
        total_samples_svm_text_ner
    )

    registros_svm_text_ner = int(cm_svm_text_ner[i, :].sum())

    rows_svm_text_ner.append({
        "Class": int(label),
        "Accuracy": builtins.round(float(accuracy_svm_text_ner_cls), 4),
        "Precision": builtins.round(float(precision_svm_text_ner), 4),
        "Recall": builtins.round(float(recall_svm_text_ner), 4),
        "F1-score": builtins.round(float(f1_svm_text_ner), 4),
        "Rows": registros_svm_text_ner
    })

# Mostrar tabla de métricas
metrics_by_class_svm_text_ner = pd.DataFrame(rows_svm_text_ner)
display(metrics_by_class_svm_text_ner)
```

```python
# Labels reales y predicciones
y_true_lstm_text = y_test_lstm_text.astype(int)
y_pred_lstm_text = y_pred_lstm_text.astype(int)

# Etiquetas únicas
labels_lstm_text = sorted(np.unique(y_true_lstm_text))

# Matriz de confusión
cm_lstm_text = confusion_matrix(
    y_true_lstm_text,
    y_pred_lstm_text,
    labels=labels_lstm_text
)

# Normalización por filas (Recall implícito)
cm_norm_lstm_text = (
    cm_lstm_text.astype("float") /
    cm_lstm_text.sum(axis=1)[:, np.newaxis]
)

# DataFrame para graficar
cm_plot_df_lstm_text = pd.DataFrame(
    cm_lstm_text,
    index=[f"Real {l}" for l in labels_lstm_text],
    columns=[f"Pred {l}" for l in labels_lstm_text]
)

# Texto (conteo + porcentaje)
text_cm_lstm_text = [
    [
        f"{cm_lstm_text[i, j]}<br>({cm_norm_lstm_text[i, j]:.1%})"
        for j in range(cm_lstm_text.shape[1])
    ]
    for i in range(cm_lstm_text.shape[0])
]

# Paleta morada – Salud Mental / Riesgo
purple_risk_scale = [
    "#f4f1fa",
    "#e0d6f5",
    "#c1a7e2",
    "#9b6bd6",
    "#6a3fa0",
    "#432874"
]

# Gráfico Plotly
fig_cm_lstm_text = px.imshow(
    cm_plot_df_lstm_text,
    text_auto=False,
    color_continuous_scale=purple_risk_scale,
    aspect="auto",
    title="Matriz de Confusión – Long Short-Term Memory (Texto)"
)

fig_cm_lstm_text.update_traces(
    text=text_cm_lstm_text,
    texttemplate="%{text}",
    textfont_size=20,
    textfont_color="white"
)

fig_cm_lstm_text.update_layout(
    title_x=0.5,
    xaxis_title="Clase Predicha",
    yaxis_title="Clase Real",
    font=dict(
        family="Algerian, DejaVu Sans, sans-serif",
        size=18
    ),
    height=600,
    width=720,
    margin=dict(l=90, r=90, t=120, b=70),
    coloraxis_colorbar=dict(title="Frecuencia")
)

fig_cm_lstm_text.show()

# Métricas por clase
rows_lstm_text = []
total_samples_lstm_text = cm_lstm_text.sum()

for i, label in enumerate(labels_lstm_text):
    TP_lstm_text = cm_lstm_text[i, i]
    FN_lstm_text = cm_lstm_text[i, :].sum() - TP_lstm_text
    FP_lstm_text = cm_lstm_text[:, i].sum() - TP_lstm_text
    TN_lstm_text = (
        total_samples_lstm_text -
        (TP_lstm_text + FP_lstm_text + FN_lstm_text)
    )

    precision_lstm_text = (
        TP_lstm_text / (TP_lstm_text + FP_lstm_text)
        if (TP_lstm_text + FP_lstm_text) > 0 else 0.0
    )

    recall_lstm_text = (
        TP_lstm_text / (TP_lstm_text + FN_lstm_text)
        if (TP_lstm_text + FN_lstm_text) > 0 else 0.0
    )

    f1_lstm_text = (
        2 * precision_lstm_text * recall_lstm_text /
        (precision_lstm_text + recall_lstm_text)
        if (precision_lstm_text + recall_lstm_text) > 0 else 0.0
    )

    accuracy_lstm_text_cls = (
        (TP_lstm_text + TN_lstm_text) /
        total_samples_lstm_text
    )

    registros_lstm_text = int(cm_lstm_text[i, :].sum())

    rows_lstm_text.append({
        "Class": int(label),
        "Accuracy": builtins.round(float(accuracy_lstm_text_cls), 4),
        "Precision": builtins.round(float(precision_lstm_text), 4),
        "Recall": builtins.round(float(recall_lstm_text), 4),
        "F1-score": builtins.round(float(f1_lstm_text), 4),
        "Rows": registros_lstm_text
    })

# Mostrar tabla de métricas
metrics_by_class_lstm_text = pd.DataFrame(rows_lstm_text)
display(metrics_by_class_lstm_text)
```

```python
# Función para obtener métricas por clase desde predicciones Spark
def metricas_por_clase_spark(df_predicciones):
    df_pd = df_predicciones.select("class", "prediction").toPandas()

    reporte = classification_report(
        df_pd["class"],
        df_pd["prediction"],
        output_dict=True,
        zero_division=0
    )

    accuracy = reporte["accuracy"]

    return {
        "0": {
            "Accuracy": accuracy,
            "Precision": reporte["0"]["precision"],
            "Recall": reporte["0"]["recall"],
            "F1-score": reporte["0"]["f1-score"]
        },
        "1": {
            "Accuracy": accuracy,
            "Precision": reporte["1"]["precision"],
            "Recall": reporte["1"]["recall"],
            "F1-score": reporte["1"]["f1-score"]
        }
    }

# Función para obtener métricas por clase desde predicciones Deep learning
def metricas_por_clase_dl(y_true, y_pred):
    reporte = classification_report(
        y_true,
        y_pred,
        output_dict=True,
        zero_division=0
    )

    accuracy = reporte["accuracy"]

    return {
        "0": {
            "Accuracy": accuracy,
            "Precision": reporte["0"]["precision"],
            "Recall": reporte["0"]["recall"],
            "F1-score": reporte["0"]["f1-score"]
        },
        "1": {
            "Accuracy": accuracy,
            "Precision": reporte["1"]["precision"],
            "Recall": reporte["1"]["recall"],
            "F1-score": reporte["1"]["f1-score"]
        }
    }

# Diccionario de modelos Spark
modelos_spark = {
    "Naive Bayes (Texto)": nb_text_predictions,
    "Naive Bayes (Texto + NER)": nb_text_ner_predictions,
    "Logistic Regression (Texto)": lr_predictions_text,
    "Logistic Regression (Texto + NER)": lr_predictions_text_ner,
    "Support Vector Machine (Texto)": svm_text_predictions,
    "Support Vector Machine (Texto + NER)": svm_ner_predictions
}

# Diccionario de modelos Deep learning
modelos_dl = {
    "Convolutional Neural Network (Texto)": (y_test_cnn_text, y_pred_cnn_text),
    "Long Short-Term Memory (Texto)": (y_test_lstm_text, y_pred_lstm_text),
    "Gated Recurrent Unit (Texto)": (y_test_gru_text, y_pred_gru_text)
}

# Inicialización de estructuras de métricas
metricas_clase_0 = {
    "Accuracy": [],
    "Precision": [],
    "Recall": [],
    "F1-score": []
}

metricas_clase_1 = {
    "Accuracy": [],
    "Precision": [],
    "Recall": [],
    "F1-score": []
}

orden_modelos = []

# Cálculo métricas modelos Spark
for modelo, df_pred in modelos_spark.items():
    metricas = metricas_por_clase_spark(df_pred)
    orden_modelos.append(modelo)

    for metrica in metricas_clase_0.keys():
        metricas_clase_0[metrica].append(metricas["0"][metrica] * 100)
        metricas_clase_1[metrica].append(metricas["1"][metrica] * 100)

# Cálculo métricas modelos Deep learning
for modelo, (y_true, y_pred) in modelos_dl.items():
    metricas = metricas_por_clase_dl(y_true, y_pred)
    orden_modelos.append(modelo)

    for metrica in metricas_clase_0.keys():
        metricas_clase_0[metrica].append(metricas["0"][metrica] * 100)
        metricas_clase_1[metrica].append(metricas["1"][metrica] * 100)

# Creación dataframes por clase
metricas_clase_0_df = pd.DataFrame(metricas_clase_0, index=orden_modelos)
metricas_clase_1_df = pd.DataFrame(metricas_clase_1, index=orden_modelos)

# Definición de escala de colores
colorscale = [
    [0.0, "#e6e6e6"],
    [0.5, "#b2ebf2"],
    [1.0, "#008080"]
]

# Creación figura con dos heatmaps
fig = make_subplots(
    rows=2,
    cols=1,
    subplot_titles=[
        "Métricas Clase Non-Suicide (0)",
        "Métricas Clase Suicide (1)"
    ],
    vertical_spacing=0.12
)

# Ajustar tamaño de los títulos de los subplots
for i, annotation in enumerate(fig['layout']['annotations']):
    annotation['font'] = dict(size=24, family='Algerian')

# Heatmap clase 0
fig.add_trace(
    go.Heatmap(
        z=metricas_clase_0_df.values,
        x=metricas_clase_0_df.columns,
        y=metricas_clase_0_df.index,
        colorscale=colorscale,
        zmin=0,
        zmax=100,
        text=metricas_clase_0_df.round(1).astype(str) + "%",
        texttemplate="%{text}",
        hovertemplate=
            "<b>Modelo:</b> %{y}<br>"
            "<b>Métrica:</b> %{x}<br>"
            "<b>Valor:</b> %{z:.1f}%<extra></extra>",
        showscale=False,
        xgap=6,
        ygap=6
    ),
    row=1,
    col=1
)

# Heatmap clase 1
fig.add_trace(
    go.Heatmap(
        z=metricas_clase_1_df.values,
        x=metricas_clase_1_df.columns,
        y=metricas_clase_1_df.index,
        colorscale=colorscale,
        zmin=0,
        zmax=100,
        text=metricas_clase_1_df.round(1).astype(str) + "%",
        texttemplate="%{text}",
        hovertemplate=
            "<b>Modelo:</b> %{y}<br>"
            "<b>Métrica:</b> %{x}<br>"
            "<b>Valor:</b> %{z:.1f}%<extra></extra>",
        showscale=False,
        xgap=6,
        ygap=6
    ),
    row=2,
    col=1
)

# Configuración general del layout
fig.update_layout(
    title=dict(
        text="Comparación X Clase Métricas Modelos Evaluados",
        x=0.5,
        xanchor="center",
        font=dict(
            family="Algerian",
            size=28
        )
    ),
    font=dict(
        family="Algerian",
        size=16
    ),
    width=1450,
    height=1200,
    margin=dict(
        t=140,
        l=220
    ),
    plot_bgcolor="white",
    paper_bgcolor="white",
    yaxis=dict(autorange="reversed"),
    yaxis2=dict(autorange="reversed")
)

# Mostrar gráfico interactivo
fig.show()
```

###**Tokenización**

En este apartado se realizará la tokenización del texto utilizando RegexTokenizer, normalizando el contenido a minúsculas y eliminando caracteres no alfabéticos. Posteriormente, se depurarán los tokens vacíos o nulos y se generará una vista preliminar del conjunto de prueba, con el fin de verificar la correcta transformación del texto antes de su uso en el modelo Convolutional Neural Network (CNN).

```python
# Librerias
import pyspark.sql.functions as F
from pyspark.ml.feature import RegexTokenizer
from IPython.core.display import display, HTML

# Tokenización con RegexTokenizer
tokenizer = RegexTokenizer(
    inputCol="Texto",
    outputCol="Tokens",
    pattern=r"[^a-zA-ZáéíóúñÁÉÍÓÚÑ]+",
    toLowercase=True,
    minTokenLength=1
)

# Crear dataframe df_clean
df_clean = tokenizer.transform(df_cnn_test_sentences)

# Eliminar tokens vacíos o nulos
df_clean = df_clean.withColumn(
    "Tokens",
    F.expr("filter(tokens, x -> x != '' AND x IS NOT NULL)")
)

# Crear preview
sample_pd = (
    df_clean.select("ID", "Texto", "Tokens", "Clase")
            .limit(10)
            .toPandas()
)

html_preview = sample_pd.to_html(escape=False, index=False)

# Mostrar tabla con scroll
display(HTML(f"""
<div style="
    max-width: 100%;
    max-height: 500px;
    overflow-x: auto;
    overflow-y: auto;
    border: 1px solid #ccc;
    padding: 15px;
    font-size: 15px;
">
{html_preview}
</div>
"""))
```

###**Eliminación Stopwords**

En este apartado se eliminarán las palabras vacías (stopwords) en inglés a partir de los tokens generados, con el objetivo de reducir ruido y conservar únicamente términos con mayor carga semántica. Posteriormente, se reorganizará la estructura del DataFrame y se mostrará una vista preliminar, permitiendo verificar el resultado del preprocesamiento antes de la evaluación del modelo Convolutional Neural Network (CNN).

```python
# Librerías
from pyspark.ml.feature import StopWordsRemover

# Stopwords en inglés
stopwords_en = StopWordsRemover.loadDefaultStopWords("english")

# Configurar y aplicar StopWordsRemover
remover = StopWordsRemover(
    inputCol="Tokens",
    outputCol="Tokens_no_stopwords",
    stopWords=stopwords_en
)

df_clean = remover.transform(df_clean)

# Reemplazar la columna Tokens por la versión sin stopwords
df_clean = df_clean.drop("Tokens").withColumnRenamed("Tokens_no_stopwords", "Tokens")

# Reordenar columnas para que "Clase" quede al final
cols = [c for c in df_clean.columns if c != "Clase"] + ["Clase"]
df_clean = df_clean.select(*cols)

# Crear preview con todas las columnas
preview_pd = df_clean.limit(10).toPandas()

# Convertir a HTML
html_preview = preview_pd.to_html(escape=False, index=False)

# Mostrar tabla con scroll
display(HTML(f"""
<div style="
    max-width: 100%;
    max-height: 550px;
    overflow-x: auto;
    overflow-y: auto;
    border: 1px solid #ccc;
    padding: 15px;
    font-size: 15px;
">
{html_preview}
</div>
"""))
```

###**Conversión Tokens Secuencias Numéricas**

En este apartado se transformarán los tokens preprocesados en secuencias numéricas mediante un Tokenizer de Keras, limitando el tamaño del vocabulario y gestionando palabras fuera de vocabulario. Posteriormente, se incorporará esta representación numérica al conjunto de datos y se reconstruirá el DataFrame en Spark, con el fin de preparar las características de entrada necesarias para la evaluación del modelo Convolutional Neural Network (CNN).

```python
# Librerias
from pyspark.sql import SparkSession
from tensorflow.keras.preprocessing.text import Tokenizer

# Convertir tokens a cadena para Tokenizer
df_pd = df_clean.select("Tokens").toPandas()
tokens_list = [" ".join(tokens) for tokens in df_pd["Tokens"]]

# Crear y entrenar tokenizer
vocab_size = 20000
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(tokens_list)

# Convertir cada fila a secuencia numérica
tokens_numeric_list = tokenizer.texts_to_sequences(tokens_list)

# Agregar la nueva columna al DataFrame pandas
df_pd["Tokens_Numerico"] = tokens_numeric_list

# Mantener todas las columnas originales en Spark
df_spark = df_clean.toPandas()
df_spark["Tokens_Numerico"] = df_pd["Tokens_Numerico"]

# Convertir de nuevo a Spark DataFrame
spark = SparkSession.builder.getOrCreate()
df_features = spark.createDataFrame(df_spark)

# Reordenar columnas para que Clase quede al final
cols = [c for c in df_features.columns if c.lower() != "clase"] + ["Clase"]
df_features = df_features.select(*cols)

# Mostrar preview
preview_pd = df_features.limit(10).toPandas()
html_preview = preview_pd.to_html(escape=False, index=False)

# Mostrar tabla con scroll
display(HTML(f"""
<div style="
    max-width: 100%;
    max-height: 400px;
    overflow-x: auto;
    overflow-y: auto;
    border: 1px solid #ccc;
    border-radius: 6px;
    padding: 10px;
">
{html_preview}
</div>
"""))
```

###**Padding Tokens Secuencias Numéricas**

Posteriormente, se integrarán las secuencias ajustadas al conjunto de datos y se reorganizarán las columnas, permitiendo verificar la estructura final de las características antes de la evaluación del modelo.

```python
# Librerias
import builtins
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Extraer la columna Tokens_Numerico a pandas
df_seq_pd = df_features.select("Tokens_Numerico").toPandas()
sequences = df_seq_pd["Tokens_Numerico"].tolist()

# Obtener la longitud máxima de las secuencias
max_len = builtins.max(len(seq) for seq in sequences) if sequences else 0
print("Longitud Máxima = ", max_len)

# Aplicar padding post
padded = pad_sequences(
    sequences,
    maxlen=max_len,
    padding="post",
    truncating="post"
)

# Reemplazar la columna Tokens_Numerico con las secuencias padded
df_seq_pd["Tokens_Numerico"] = padded.tolist()

# Traer el resto de columnas a pandas
other_cols = [c for c in df_features.columns if c != "Tokens_Numerico"]
df_other_pd = df_features.select(other_cols).toPandas()

# Insertar la columna padded como Tokens_Numerico
df_other_pd.insert(df_other_pd.columns.get_loc("Tokens") + 1, "Tokens_Numerico", df_seq_pd["Tokens_Numerico"])

# Convertir de nuevo a Spark DataFrame
spark = SparkSession.builder.getOrCreate()
df_features = spark.createDataFrame(df_other_pd)

# Reordenar columnas exactamente como quieres
desired_order = ["ID", "Texto", "Tokens", "Tokens_Numerico", "Clase"]
df_features = df_features.select(*desired_order)

# Crear preview en HTML
preview_pd = df_features.limit(10).toPandas()
html_preview = preview_pd.to_html(escape=False, index=False)

# Mostrar tabla con scroll
display(HTML(f"""
<div style="
    max-width: 100%;
    max-height: 500px;
    overflow-x: auto;
    overflow-y: auto;
    border: 1px solid #ccc;
    border-radius: 8px;
    padding: 12px;
    box-shadow: 0px 2px 8px rgba(0,0,0,0.08);
    font-size: 13px;
">
{html_preview}
</div>
"""))
```

###**Matriz Confusión Convolutional Neural Network (CNN)**
